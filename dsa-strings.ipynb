Find the first non-repeating character.

```python
strings="loveleetcode"
```

```python
# with count function
hold={}
non_repet,repet=[],[]
for i in strings:
    hold[i]=strings.count(i)
    non_repet.append(i) if hold[i]<=1 else repet.append(i)
rep=""
non_rep=""
for i in repet:
    rep+=i
for i in non_repet:
    non_rep+=i
print("repeat:",rep)
print("not_repeat:",non_rep)          
```

```python
total=0
word=""
for i in strings_1:
    if len(i)>total:
        word=i
        total=len(i)
print(word)
    
```

Find the shortest word in a sentence.

```python
totals=0
words=''
for i in strings_1:
    total_2=len(i)
    word=i
    if total_2<totals or totals==0:
        totals=total_2
        words=word
print(f"shortest word in this sentence is [{words}]")
```

Check if a string contains only digits.

```python
strings_1='12345'
a=''
for i in strings_1:
    try:
        if type(int(i))==int:
            a='yes'
    except ValueError:
        a='no'
        break
print(a)
```
