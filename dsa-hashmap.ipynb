HASHMAP

Find the frequency of elements using HashMap.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    print(f"number {i} appears {hold[i]} times")
    if hold[i]==1:
        print(f"{i} appears 1 times")

```

Find the first non-repeating element.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    if hold[i]==1:
        print(f"{i} appears 1 times")
        break
```

Find the first repeating element.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    if hold[i]>1:
        print(f"first repeting number is {i} ")
        break
```

Find duplicate elements using HashMap.

```python
list_2=[1,2,3,4,5,6,7,6,3,1,3]
duplicate=[]
for i in list_2:
    if i not in duplicate:
        if list_2.count(i)>1:
            duplicate.append(i)
print(duplicate)
```

Find the most frequent element.

```python
lst_1=["apple","banana","apple","banana","apple","cherry"]
dic_1={}
high,high_name=0,""
for i in lst_1:
    if i not in dic_1:
        num=lst_1.count(i)
        dic_1[i]=num
        if high<num:
            high,high_name=num,i
print(f"{high_name} appears high count of {high}")
```

Check if two arrays contain the same elements.

```python
list_1=[1,2,3,4,5]
list_2=[5,4,3,2,1]
list_1=set(list_1)
list_2=set(list_2)
print(list_1)
print(list_2)
output=True
for i in list_1:
    if i not in list_2:
        output=False
print(output)
```

Check if two strings are anagrams using HashMap.

```python
strings_1="malayalam"
strings_2="malayalam"
output=True
strings_1={strings_1}
strings_2={strings_2}
for i in strings_1:
    if len(strings_1)==len(strings_2):
        if i not in strings_2:
            output=False
            break
print(output)
```
