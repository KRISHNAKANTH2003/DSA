Find the frequency of elements using HashMap.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    print(f"number {i} appears {hold[i]} times")
    if hold[i]==1:
        print(f"{i} appears 1 times")

```

Find duplicate elements using HashMap.

```python
list_2=[1,2,3,4,5,6,7,6,3,1,3]
duplicate=[]
for i in list_2:
    if i not in duplicate:
        if list_2.count(i)>1:
            duplicate.append(i)
print(duplicate)
```

Find the most frequent element.

```python
lst_1=["apple","banana","apple","banana","apple","cherry"]
dic_1={}
high,high_name=0,""
for i in lst_1:
    if i not in dic_1:
        num=lst_1.count(i)
        dic_1[i]=num
        if high<num:
            high,high_name=num,i
print(f"{high_name} appears high count of {high}")
```

Check if two arrays contain the same elements.

```python
list_1=[1,2,3,4,5]
list_2=[5,4,3,2,1]
list_1=set(list_1)
list_2=set(list_2)
print(list_1)
print(list_2)
output=True
for i in list_1:
    if i not in list_2:
        output=False
print(output)
```

Group anagrams together.

```python
strings="malayalam"
right=len(strings)-1
is_that="anagram"
for i in strings:
    if i==strings[right]:
        right-=1
    else:
        is_that="not anagram"
        break
print(is_that)
```

```python
def dic(a:str):
    dic={}
    for i in range(len(a)):
        dic[i]=a[i]
    return dic

```

```python
strings=dic("silent")
strings_2=dic('listen')
```

```python
print(strings)
print(strings_2)
output='group anagram'
if len(strings.values())==len(strings_2.values()):
    for i in strings.values():
        if  i not in strings_2.values():
            output='not anagram'
            break
else:
    output='not anagram'
print(output)
```

Find the longest substring without repeating characters.

```python
strings="ababacddabc"
hold=[]
left,right=0,1
first_total,total=0,0
for i in range(len(strings)):
    if strings[i] not in hold:
        hold.append(strings[i])
        total+=1
        right+=1
    else:
        if total>first_total:
            left+=1
            first_total=total
        else:
            left+=1
print(first_total)
```

Find the character with the highest frequency.

```python
strings="krishnakanthk"
test_strings=set([i for i in strings])
total,char=0,""
for i in test_strings:
    if strings.count(i)>total:
        total,char=strings.count(i),i
print(f"letter {char} has {total} frequency")
```

Count the frequency of words in a sentence.

```python
strings="apple banana apple orange banana apple".split(" ")
total_freq,word=0,""
for i in strings:
    word_count=strings.count(i)
    if word_count>total_freq:
        total_freq,word=word_count,i
print(f"{word} has high frequency count of {total_freq}")
```
