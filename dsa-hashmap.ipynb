HASHMAP

Find the frequency of elements using HashMap.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    print(f"number {i} appears {hold[i]} times")
    if hold[i]==1:
        print(f"{i} appears 1 times")

```

Find the first non-repeating element.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    if hold[i]==1:
        print(f"{i} appears 1 times")
        break
```

Find the first repeating element.

```python
list_1=[1,2,3,4,5,4,3,2,3,2,3,2,3,4,5]
hold={}
for i in list_1:
    if i in hold.keys():
        hold[i]=hold[i]+1  
    else:
        hold[i]=1
for i in hold:
    if hold[i]>1:
        print(f"first repeting number is {i} ")
        break
```

Find duplicate elements using HashMap.

```python
list_2=[1,2,3,4,5,6,7,6,3,1,3]
duplicate=[]
for i in list_1:
    if i not in duplicate:
        if list_2.count(i)>1:
            duplicate.append(i)
print(duplicate)
```
